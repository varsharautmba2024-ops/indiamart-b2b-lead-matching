```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Seller database
sellers = pd.DataFrame({
    "Seller": [
        "ABC Safety",
        "XYZ Furniture",
        "PQR Machinery",
        "SafePro",
        "BuildTech"
    ],
    "Description": [
        "Industrial safety helmets gloves protective equipment",
        "Office chairs desks office furniture",
        "Construction machinery drilling machines equipment",
        "Safety helmets jackets gloves industrial PPE",
        "Cement mixers construction equipment tools"
    ]
})

# Buyer requirement
buyer_requirement = "bulk industrial safety helmets and protective equipment"

# Convert text into numerical vectors
vectorizer = TfidfVectorizer(stop_words="english")
texts = sellers["Description"].tolist() + [buyer_requirement]
vectors = vectorizer.fit_transform(texts)

# Calculate similarity
similarities = cosine_similarity(
    vectors[-1],
    vectors[:-1]
).flatten()

# Rank sellers
sellers["Relevance Score"] = similarities
result = sellers.sort_values("Relevance Score", ascending=False)

print("Buyer Requirement:")
print(buyer_requirement)
print("\nRanked Sellers:")
print(result.to_string(index=False))
```
